##### *1.2 Funcion para aplicar algoritmo de de White Patch*

```python
# Por defecto se considera el percentile 99
def white_patch(img, percentile=99):
    img = img.astype(np.float64)
    max_value = np.percentile(img, percentile) #calculamos el maximo de cada canal
    img_res = img * (255 /max_value) 
    img_res = np.clip(img_res, 0, 255).astype(np.uint8)   
    
    return img_res
```

```python
def plot_white_patch(img,imgwp1, imgwp2, imgwp3):
    fig = plt.figure(figsize=(15,10))

    ax1 = fig.add_subplot(141)
    ax1.imshow(img, cmap='gray')
    ax1.set_title("Imagen original")

    ax2 = fig.add_subplot(142)
    ax2.imshow(imgwp1, cmap='gray')
    ax2.set_title("Imagen corregida con perc=99")

    ax3 = fig.add_subplot(143)
    ax3.imshow(imgwp2, cmap='gray')
    ax3.set_title("Imagen corregida con perc=97")

    ax4 = fig.add_subplot(144)
    ax4.imshow(imgwp3, cmap='gray')
    ax4.set_title("Imagen corregida con perc=95")

    plt.tight_layout()
    plt.show()
```

##### *1.2 Correccion de iluminación con algoritmo de White Path*

```python
imgwp_p99_test_b = white_patch(img_test_b)
imgwp_p97_test_b = white_patch(img_test_b, percentile=97)
imgwp_p95_test_b = white_patch(img_test_b, percentile=95)

plot_white_patch(img_test_b, imgwp_p99_test_b, imgwp_p97_test_b, imgwp_p95_test_b)
```

**Analisis del algorimo**  
Aplicar el algoritmo de White Patch sobre imagenes en escala de grises no realiza una corrección del balance de color dado que no hay canales por comparar entre si, por lo tanto, sería mas un algoritmo de estiramiento o reescalado del contraste usando el percentil como referencia de blanco o como el nuevo blanco.
Asimismo, al utilizar un percentil menor a 100 nos permite obtener un estiramiento del contraste mucho mas efectivo debido a que, si usamos el maximo absoluto (perc=100) y este es un outlier saturado el algoritmo practicamente no corrige nada porque asume que la imagen ya tiene blanco, mientras que al reducir el percentil se ignoran estos outliers. Sin embargo, si reducimos demasiado el percentil se empiezan a saturar zonas que si contienen información válida.

Esto podria verse como una limitacion clara del algoritmo, que usa como referencia el blanco pero no manipula el extremo oscuro, por lo tanto no se podria recuperar algun detalle en zonas que se han puesto totalmente negras y han perdido información.

En conclusión, cuanto mas lejos está el percentil elegido del valor maximo real de brillo de la imagen, mas agresiva sera la correción y podriamos saturar zonas que ya son claras y por el contrario, cuanto mas cerca se encuentra la correción es menor. Para todos los casos, las sombras intensas quedan sin corregir, dado que el algoritmo trabaja tomando como referencia el blanco.
 


#### *2. Histogramas*

##### *2.1 Lectura de las imagenes en escala de grises*

```python
img1 = cv2.imread('img1_tp.png',cv2.IMREAD_GRAYSCALE)
InfoImg(img1)
```

```python
img2 = cv2.imread('img2_tp.png',cv2.IMREAD_GRAYSCALE)
InfoImg(img2)
```

##### *2.2 Visualización de imagenes*

```python
fig = plt.figure(figsize=(20,12))

ax1 = fig.add_subplot(121)
ax1.imshow(img1, cmap='gray')
ax1.set_title("Imagen 1")

ax2 = fig.add_subplot(122)
ax2.imshow(img2, cmap='gray')
ax2.set_title("Imagen 2")

plt.tight_layout()
plt.show()
```

##### *2.3 Histogramas*

```python
# Funcion: cv2.calcHist(images, channels, mask, histSize, ranges[]]
n_bins = 32
hist_img1 = cv2.calcHist([img1],[0],None,[n_bins],[0,256])

fig = plt.figure(figsize=(20,12))

ax1 = fig.add_subplot(121)
ax1.imshow(img1, cmap='gray')
ax1.set_title("Imagen 1")

ax2 = fig.add_subplot(122)
plt.fill_between(range(n_bins), hist_img1.ravel(), color='b', alpha=0.3)
ax2.plot(hist_img1, color='b', linewidth=1)
ax2.set_title("Histograma Imagen 1")
plt.ylabel('Frecuencia relativa')

plt.tight_layout()
plt.show()
```

```python
hist_img2 = cv2.calcHist([img2],[0],None,[n_bins],[0,256])

fig = plt.figure(figsize=(20,12))

ax1 = fig.add_subplot(121)
ax1.imshow(img2, cmap='gray')
ax1.set_title("Imagen 2")

ax2 = fig.add_subplot(122)
plt.fill_between(range(n_bins), hist_img2.ravel(), color='b', alpha=0.3)
ax2.plot(hist_img2, color='b', linewidth=1)
ax2.set_title("Histograma Imagen 2")
plt.ylabel('Frecuencia relativa')

plt.tight_layout()
plt.show()
```

##### *2.5 Comparación de histogramas*

```python
# Normalizamos los histogramas para comparar independientemente del tamano de imagen
hist1_norm = hist_img1 / hist_img1.sum()
hist2_norm = hist_img2 / hist_img2.sum()
 
bin_edges = np.linspace(0, 256, n_bins + 1) #generamos los limites de cada bin
bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2 #calculamos el centro de cada bin    

# Histograma superpuesto para comparacion directa
plt.figure(figsize=(8, 5))
plt.plot(bin_centers, hist1_norm, marker='o', label='img1_tp.png', color='steelblue')
plt.plot(bin_centers, hist2_norm, marker='o', label='img2_tp.png', color='darkorange')
plt.title(f'Comparacion de histogramas ({n_bins} bins)')
plt.xlabel('Nivel de gris')
plt.ylabel('Frecuencia relativa')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
```

##### *2.6 Análisis de histogramas*

**Selección del número de bins**  
Elegimos 32 bins para obtener un histograma mucho mas suavizado y desde el punto de vista estadistico mas robusto. Esto debido a que al tener 82,656 px en total, con 256 bins cada bin agruparia en promedio 320 px lo que muestra un histograma con una varianza alta o visualmente con muchos dientes de sierra, por el contrario con 32 bins, cada uno agruparia en promedio 2,500 px lo que hace que sea mas robusto obteniendo un balance entre resolución y suavidad, ademas que para la comparación de los histogramas seria una ventaja ya que reduce los efectos de pequeños corrimientos de intensidad entre imagenes (por ejemplo tener dos pixeles de valores 22 y 23 estarian en el mismo bin en vez de considerarse en bins diferentes).

**Histogramas de cada imagen**  
Al observar la imagen 1 podemos ver un degradado que aparentemente barre todos los tonos de gris de forma lineal y constante, pero al ver el histograma podemos ver una distribucion bimodal con dos picos y un valle que la velocidad de cambio del degradado no es constante dado que hay tramos donde el gris cambia lentamente fila a fila lo que genera el pico en el histograma y tramos donde cambia rápido lo que genera el valle. Por otro lado con respecto al histograma de la figura 2 observamos igual una distribución bimodal con una gran acumulación de pixeles oscuros que corresponden al fondo de la flor y otra acumulación de pixeles claros que corresponden a los pétalos de la flor. 

Al realizar la comparativa de ambos histogramas vemos que ambos son identicos dado que tienen los mismos tonos de gris, mismas cantidades de cada tono, pero distribuidos espacialmente de forma completamente distinta. En este punto podemos notar que el histograma como descriptor presenta la desventaja de ser invariante a la posición de los pixeles o sea, solo cuenta la cantidad de pixeles por intensidad sin importar donde esten ubicados, por eso la imagen del degradado puede tener la misma distribución de intensidades que la flor que contiene bordes y de por si mas detalles.

**¿Puede ser de utilidad tomar como ‘features’ a los histogramas en un problema de clasificación/detección de imagenes?**  
Con el ejemplo de las dos imagenes que tienen contenidos distintos pero con el mismo histogramas, por si solos los histogramas no serían suficientes para clasificarlas, pero si se combinan con otros atributos como la posición de los pixeles, la intensidad de los pixeles, etc. creo que se puede obtener una imagen clasificada de forma correcta.
